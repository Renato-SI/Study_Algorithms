## Tempo de Execução

Sempre que falarmos sobre um algoritmo, estaremos nos referindo ao seu **tempo de execução**. Geralmente, você escolhe o mais eficiente caso esteja querendo otimizar tempo e espaço.

Voltando para o contexto da **pesquisa simples x binária**, quando comparamos esses dois algoritmos, vemos que, por um lado, a pesquisa simples precisa de **4 bilhões de tentativas** para acertar um número dentre os 4 bilhões. Isso é chamado de **Tempo Linear**.

Já a pesquisa binária é diferente. Com os mesmos **4 bilhões de itens**, você precisaria de no máximo **32 tentativas**. Uma notável diferença, não? A pesquisa binária é executada em **Tempo Logarítmico**.

## Notação Big O

A notação **Big O** é uma notação especial que diz o quão rápido é um algoritmo, baseado na quantidade de operações.

Aqui vai uma comparação:

Bob tem apenas **10 segundos** para descobrir onde o foguete da NASA deve pousar. Ele está em dúvida entre a pesquisa binária e a simples. Lembrando que ele não pode tolerar erros!

O raciocínio de Bob foi o seguinte:

> "Se para 100 elementos a pesquisa binária (7 ms) é 15x mais rápida do que a simples (30 ms), então com 1 bilhão a lógica deve se manter. Logo, 30 × 15 = 450 ms, correto? Então irei escolher a pesquisa simples, pois terei tempo suficiente e terei uma resposta extremamente precisa!"

![Captura de tela 2026-08-20 144907.png](<attachment:Captura de tela 2026-08-20 144907.png>)

Mas a lista de locais contém **1 bilhão de elementos**. Uma pesquisa binária levaria cerca de **30 ms** para encontrar a resposta, enquanto a simples levaria **1 bilhão de ms**, ou seja, aproximadamente **11 dias**.

Então Bob definitivamente não entendia sobre seu algoritmo, pois a pesquisa binária cresce **logaritmicamente**, enquanto a simples cresce **linearmente**.

É aí que entra a **notação Big O**.

A notação Big O informa o quão rápido é um algoritmo. Imagine uma lista de **n elementos**. O tempo de execução em Big O é **O(n)**.

O Big O não fornece o tempo em segundos. A notação Big O permite que você compare o **número de operações**. Ela informa o quão rápido um algoritmo cresce.

---

### Vendo diferentes tempos de execução Big O

**Exemplo:** a tarefa é desenhar uma grade com **16 divisões**.

**Algoritmo 01:** desenha 1 divisão por vez. Logo, vai precisar de **16 etapas** para desenhar as 16 divisões. O tempo de execução deste algoritmo é **O(n)**.

**Algoritmo 02:** neste exemplo, dobrar o papel ao meio uma vez representa **1 operação**. Perceba que, a cada dobra que você faz, o número de divisões duplica, porém consumindo apenas uma operação por dobra. O tempo de execução deste algoritmo é **O(log n)**.

### A notação Big O estabelece o tempo de execução para a pior hipótese

A pesquisa simples tem o tempo de execução de **O(n)**. Porém, em um caso atípico, você precisa encontrar o elemento 1 dentre 100 elementos. O tempo de execução mudou de **O(n) para O(1)**?

A resposta é **não**, pois o Big O leva em conta o **pior caso**, aquele em que você analisou cada item da lista.

Então, o tempo **O(n)** é uma garantia. Logo, você sabe que a pesquisa simples nunca terá um tempo de execução pior que **O(n)**.

---

### Exemplos de tempos de execução Big O

1. **O(log n)** — Também conhecido como tempo logarítmico. Exemplo: **pesquisa binária**.
2. **O(n)** — Exemplo: **pesquisa simples**, tempo linear.
3. **O(n × log n)** — Exemplo de um algoritmo rápido de ordenação, como o **Quicksort**.
4. **O(n²)** — Exemplo de um algoritmo lento de ordenação.
5. **O(n!)** — Exemplo de um algoritmo **extremamente lento**, como o **Caixeiro-Viajante**.

## Considerações Finais

* A rapidez de um algoritmo não é medida em segundos, mas pelo **crescimento do número de operações**.
* O tempo de execução de um algoritmo é expresso em **Big O**.
* Deve-se analisar o quão rapidamente o tempo de execução de um algoritmo aumenta conforme o **número de elementos aumenta**.
