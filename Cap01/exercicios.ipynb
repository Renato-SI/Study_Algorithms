# 🔎 Exercícios de Pesquisa Binária

### 1.1 — Lista com 128 nomes

**Pergunta:**
Suponha que você tenha uma lista com **128 nomes** e esteja fazendo uma pesquisa binária. Qual seria o número máximo de etapas que você levaria para encontrar o nome desejado?

**Minha resposta:**

Partindo-se do princípio de que o algoritmo da **pesquisa binária elimina metade das possibilidades a cada "chute"**, posso concluir que o número máximo de etapas é de **log₂ n**, ou seja:

> **log₂ 128 = 7!**

Então, para chegar em apenas **um elemento** e, consequentemente, encontrar com precisão o nome na lista, eu levaria **07 etapas**!

---

### 1.2 — Dobrando o tamanho da lista

**Pergunta:**
Suponha que você **duplique o tamanho da lista**. Qual seria o número máximo de etapas agora?

**Minha resposta:**

Seguindo a mesma lógica anterior, **log₂ n**, consigo rapidamente concluir que:

> **log₂ 256 = 8!**

Então, para uma lista com **256 nomes**, eu levaria no máximo **8 etapas**.
