## Pesquisa Binária

Suponha que você esteja procurando o nome de uma pessoa em uma agenda telefônica, no qual o nome começa com K. Você pode começar da primeira página e ir folheando até chegar aos Ks, porém você provavelmente vai começar pela metade, pois sabe que os Ks estarão mais perto dali.

O mesmo exemplo acima se aplica para outras questões como: Imagine que o Facebook precissa verificar se um determinado usuário existe ou não. Digamos que o usuário seja "Renatop1w1". O facebook poderia começar a procurá em seu enorme banco de dados pelos As - mas faz mais sentido que ele comece a buscar do meio.

**Isto é um problema de busca.** E todos estes casos usam um algoritmo para resolvê-lo da melhor forma possível: **PESQUISA BINÁRIA!**
A pesquisa Binária é um algoritmo. Sua entrada é uma lista **Ordenada** de n elementos. Se o elemento qwue você está buscando está na lista ela retorna a sua localização. Caso contrário (Se não estiver) ela retorna **None**.

Eis um exemplo de como a pesquisa Binária Funciona. Estou pensando em um número entre 1 e 100.



Você deve procurar adivinhar o meu número com o menor número de tentativas possível. A cada tentativa direi se você chutou acima ou abaixo ou corretamente o número. 

Digamos que você tentou assim: 1,2,3,4... Observe como seria.



Isso se chama pesquina simples (Pesquisa burra). A cada tentativa vc está eliminando apenas um número e isso é Péssimo pensando em uma **""Solução Otimizada"** porque você precisária de 99 chutes para acertar com certeza!

## **Uma maneira melhor de buscar**

Uma técnica melhor seria começar pelo 50, assim você eliminaria metade da Lista de uma vez só!



Agora o seu próximo chute deveria ser 75, se a resposta fosse: Muito alto!
agora você sabe que o número que vc busca está entre 50 e 75. Então seu próximo chute deveria ser 63 se novamente fosse: Alto demais! você poderia tentar 57 e essa seria a resposta nesse caso!



Isso é a pesquisa Binária. Você provavelmente já utilizou este algoritmo na sua vida em várias situações e nem se deu conta! Aqui está a quantidade de números que você pode eliminar a cada tentativa.



Seja qual for o número você poderia eliminá-lo em apenas 7 tentativa. Isso ocorré porque na pesquisa Binária, a cada etapa você elimina o número de palavras pela metade até que reste apenas 1. Se Fosse 240.000 palavras e você precisa encontrar apenas 01, com a pesquisa simples você precisária de 240.000 estapas enquanto na binária de apenas 18.

De maneira geral, para uma lista de n números, a pesquisa binária precisa de log2 n para retornar o valor correto, enquanto a simples precisa de n. Normalmente sempre utilizaremos a Base 2 do log para resolver problemas.



```python
def pesquisa_binaria(lista = list, item = int):
    baixo = 0
    alto = len(lista) - 1

    while baixo <= alto:
        meio = (baixo + alto) // 2
        chute = lista[meio]
        if chute == item:
            return meio
        if chute > item:
            alto = meio - 1
        else:
            baixo = meio + 1
    return None

minha_lista = [1, 3, 5, 7, 9]

print(pesquisa_binaria(minha_lista, 9)) # => 1
print(pesquisa_binaria(minha_lista, 0)) # => None, o 0 NÃO está na lista.
```
