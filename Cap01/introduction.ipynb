# Capítulo 01

### Introdução a algoritmos

Um **algoritmo** é o **conjunto de intruções** que servem para a realização de uma tarefa. Cada trecho de um código poderia ser chamado de algoritmo, mas o **livro trata apenas dos trechos mais interessantes**, seja por serem algoritmos **rápidos** ou por resolverem problemas interessantes, ou ambos!
